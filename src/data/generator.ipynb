```python
import pandas as pd
import random
```

```python
df = pd.read_csv('RELIANCE.csv')
```

```python
df.head()
```

```python
PRICE_COL = 'Close'
QTY_COL   = 'Volume'
```

```python
simulation_rows = []
active_order_ids = []
order_id_counter = 1
```

```python
for _, row in df.iterrows():
    # Get price and quantity from the current tick
    price = float(row[PRICE_COL])
    qty = int(row[QTY_COL])
    
    # 20% chance to emit a CANCEL command if we have active orders
    if active_order_ids and random.random() < 0.20:
        cancel_id = random.choice(active_order_ids)
        active_order_ids.remove(cancel_id)
        simulation_rows.append({
            'Command': f"CANCEL,{cancel_id}"
        })
    
    # Generate an ADD command
    else :
        side = random.choice(['BUY', 'SELL'])  # Alternate or randomize side
        current_id = order_id_counter
        order_id_counter += 1
        active_order_ids.append(current_id)
        
        simulation_rows.append({
            'Command': f"ADD,{current_id},{side},{qty},{price:.2f}"
        })

sim_df = pd.DataFrame(simulation_rows)

sim_df.to_csv('simulation_input.csv', index=False, header=False)
```
